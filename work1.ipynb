```python
#import data
import pandas as pd
%matplotlib inline
```

```python
#connect and understand data
titanic = pd.read_csv("dataset/titanic.csv")
print(titanic.shape)
titanic.info
```

```python

#increase max column
pd.set_option("display.max_columns" ,50)
titanic.head(10)
```

```python
titanic.dtypes
```

```python
#exercise 1.1
#set datatype
titanic['passenger_id'] = titanic['passenger_id'].astype('string')
#set uniques
titanic.passenger_id.unique()
```

```python
#exercise 1.2
#change factor
titanic['embarked'] = pd.Categorical(
    titanic['embarked'].replace({
        'C': 'Cherbourg',
        'Q': 'Queenstown',
        'S': 'Southampton'
    }),
    categories=['Cherbourg', 'Queenstown', 'Southampton']
)

titanic['sex'] = pd.Categorical(
    titanic['sex'].replace({
        1: 'Female',
        0: 'Male'
    }),
    categories=['Female', 'Male']
)
titanic['survived'] = pd.Categorical(
    titanic['survived'].replace({
        1: 'Yes',
        0: 'No'
    }),
    categories=['Yes', 'No']
)

titanic['pclass'] = pd.Categorical(
    titanic['pclass'].replace({
        1: '1st',
        2: '2nd',
        3: '3rd'
    }),
    categories=['1st', '2nd', '3rd'],
    ordered=True
)
```
